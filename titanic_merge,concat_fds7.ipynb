```python
df1 = df.iloc[:300]
df2 = df.iloc[300:600]

df_concat = pd.concat([df1, df2], axis=0)

print("Original dataset shape:", df.shape)
print("After concat shape:", df_concat.shape)

print(df_concat.head())
```

```python
part1 = df[['PassengerId', 'Name', 'Sex']]
part2 = df[['Age', 'Fare', 'Survived']]

df_concat_columns = pd.concat([part1, part2], axis=1)

print(df_concat_columns.head())
```

```python
df_passenger = df[['PassengerId', 'Name', 'Sex']]

df_survival = df[['PassengerId', 'Survived', 'Fare']]

print(df_passenger.head())
print(df_survival.head())
```

```python
df_merge = pd.merge(
    df_passenger,
    df_survival,
    on='PassengerId'
)

print(df_merge.head())
```

```python
inner_merge = pd.merge(
    df_passenger,
    df_survival,
    on='PassengerId',
    how='inner'
)

print(inner_merge.head())
```

```python
left_merge = pd.merge(
    df_passenger,
    df_survival,
    on='PassengerId',
    how='left'
)

print(left_merge.head())
```

```python
right_merge = pd.merge(
    df_passenger,
    df_survival,
    on='PassengerId',
    how='right'
)

print(right_merge.head())
```

```python
outer_merge = pd.merge(
    df_passenger,
    df_survival,
    on='PassengerId',
    how='outer'
)

print(outer_merge.head())
```
