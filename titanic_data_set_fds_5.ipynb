```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Read Titanic dataset
df = pd.read_csv('/content/Titanic-Dataset (1).csv')

# Display first 5 rows
df.head()
```

```python
print("Shape of dataset:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nStatistical summary:")
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
# Age
age = df['Age'].dropna()

age_range = age.max() - age.min()

print("Age Range:", age_range)

# Fare
fare = df['Fare'].dropna()

fare_range = fare.max() - fare.min()

print("Fare Range:", fare_range)
```

```python
print("Age Variance:", age.var())
print("Fare Variance:", fare.var())
```

```python
print("Age Standard Deviation:", age.std())
print("Fare Standard Deviation:", fare.std())
```

```python
# Age
Q1_age = age.quantile(0.25)
Q3_age = age.quantile(0.75)
IQR_age = Q3_age - Q1_age

print("Age Q1:", Q1_age)
print("Age Q3:", Q3_age)
print("Age IQR:", IQR_age)

# Fare
Q1_fare = fare.quantile(0.25)
Q3_fare = fare.quantile(0.75)
IQR_fare = Q3_fare - Q1_fare

print("\nFare Q1:", Q1_fare)
print("Fare Q3:", Q3_fare)
print("Fare IQR:", IQR_fare)
```

```python
age_cv = (age.std() / age.mean()) * 100
fare_cv = (fare.std() / fare.mean()) * 100

print("Age Coefficient of Variation:", age_cv, "%")
print("Fare Coefficient of Variation:", fare_cv, "%")
```

```python
print("Age Skewness:", age.skew())
print("Fare Skewness:", fare.skew())
```

```python
plt.figure(figsize=(8,5))
plt.hist(age, bins=20, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Histogram of Age')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=age)
plt.xlabel('Age')
plt.title('Box Plot of Age')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=fare)
plt.xlabel('Fare')
plt.title('Box Plot of Fare')
plt.show()
```
