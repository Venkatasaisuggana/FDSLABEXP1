```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Sai', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['IT', 'HR', 'IT', 'Sales'],
    'Salary': [30000, 25000, 35000, 28000]
})

print("DataFrame 1:")
print(df1)
```

```python
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'City': ['Hyderabad', 'Chennai', 'Bangalore', 'Delhi'],
    'Experience': [2, 3, 5, 1]
})

print("\nDataFrame 2:")
print(df2)
```

```python
data = pd.Series(
    [30000, 25000, 35000, 28000, 40000],
    index=[
        ['IT', 'IT', 'HR', 'HR', 'Sales'],
        ['Sai', 'Ravi', 'Priya', 'Kiran', 'Arun']
    ]
)

print("Hierarchical Indexing:")
print(data)

print("\nSelect IT Department:")
print(data.loc['IT'])

print("\nSelect Sai's Salary:")
print(data.loc[('IT', 'Sai')])
```

```python
df_a = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['Sai', 'Ravi']
})

df_b = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['Priya', 'Kiran']
})

# Combine row-wise
row_combined = pd.concat([df_a, df_b], ignore_index=True)

print("Row-wise Combination:")
print(row_combined)

# Combine column-wise
column_combined = pd.concat(
    [df_a, df_b],
    axis=1
)

print("\nColumn-wise Combination:")
print(column_combined)
```

```python
# Inner join
inner_merge = pd.merge(df1, df2, on='ID', how='inner')
print("Inner Merge:")
print(inner_merge)
```

```python
# Left join
left_merge = pd.merge(df1, df2, on='ID', how='left')
print("\nLeft Merge:")
print(left_merge)
```

```python
# Right join
right_merge = pd.merge(df1, df2, on='ID', how='right')
print("\nRight Merge:")
print(right_merge)
```

```python
# Outer join
outer_merge = pd.merge(df1, df2, on='ID', how='outer')
print("\nOuter Merge:")
print(outer_merge)
```

```python
data = pd.DataFrame(
    np.arange(6).reshape(2, 3),
    index=pd.Index(['Ohio', 'Colorado'], name='State'),
    columns=pd.Index(['One', 'Two', 'Three'], name='Number')
)

print("Original DataFrame:")
print(data)
```

```python
stacked = data.stack()
print("\nStacked Data:")
print(stacked)
```

```python
unstacked = stacked.unstack()
print("\nUnstacked Data:")
print(unstacked)
```

```python
sales = pd.DataFrame({
    'Date': [
        '2026-01-01', '2026-01-01',
        '2026-01-02', '2026-01-02'
    ],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile'],
    'Sales': [50000, 20000, 60000, 25000]
})

print("Original Sales Data:")
print(sales)

pivot_table = sales.pivot(
    index='Date',
    columns='Product',
    values='Sales'
)

print("\nPivoted Data:")
print(pivot_table)
```

```python
wide_data = pd.DataFrame({
    'Name': ['Sai', 'Ravi', 'Priya'],
    'Maths': [85, 90, 88],
    'Science': [80, 92, 86]
})

print("Wide Format:")
print(wide_data)

long_data = pd.melt(
    wide_data,
    id_vars=['Name'],
    var_name='Subject',
    value_name='Marks'
)

print("\nLong Format:")
print(long_data)
```

```python
grouped_data = df1.groupby('Department')['Salary'].agg(
    ['mean', 'sum', 'count']
)

print("Department-wise Salary Summary:")
print(grouped_data)
```

```python
print("Missing Values:")
print(outer_merge.isnull().sum())

cleaned_data = outer_merge.copy()

cleaned_data['City'] = cleaned_data['City'].fillna('Unknown')
cleaned_data['Experience'] = cleaned_data['Experience'].fillna(0)

print("\nData After Handling Missing Values:")
print(cleaned_data)
```
