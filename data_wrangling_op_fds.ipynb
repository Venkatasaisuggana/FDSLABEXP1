```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/Titanic-Dataset.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of rows and columns:")
print(df.shape)
```

```python
print("\nColumn names:")
print(df.columns)
```

```python
print("\nDataset information:")
df.info()
```

```python
print("Last 5 records:")
display(df.tail())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)
```

```python
df_sorted = df.sort_values(by="Age")

display(df_sorted[["Name", "Age", "Sex"]].head(10))
```

```python
df = df.rename(columns={
    "PassengerId": "Passenger_ID",
    "Pclass": "Passenger_Class"
})

print(df.columns)
```

```python
age_by_class = df.groupby("Passenger_Class")["Age"].mean()

print(age_by_class)
```

```python
survival_by_class = df.groupby("Passenger_Class")["Survived"].mean()

print(survival_by_class)
```

```python
survival_percentage = df.groupby("Passenger_Class")["Survived"].mean() * 100

print(survival_percentage)
```

```python
gender_survival = df.groupby("Sex")["Survived"].mean() * 100

print("Survival percentage by gender:")
print(gender_survival)
```

```python
df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})

print(df[["Sex"]].head())
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["Fare"] < lower_limit) |
    (df["Fare"] > upper_limit)
]

print("Number of Fare outliers:", len(outliers))

display(outliers.head())
```

```python
df.to_csv("Titanic_Wrangled.csv", index=False)

print("Wrangled Titanic dataset saved successfully!")
```
